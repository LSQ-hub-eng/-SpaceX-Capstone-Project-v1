```python
%%sql
SELECT BOOSTER_VERSION, LAUNCH_SITE, PAYLOAD_MASS__KG__
FROM SPACEXTBL
WHERE LANDING_OUTCOME LIKE '%Success%';
```

```python
%%sql
SELECT DISTINCT BOOSTER_VERSION
FROM SPACEXTBL
WHERE PAYLOAD_MASS__KG__ = (SELECT MAX(PAYLOAD_MASS__KG__) FROM SPACEXTBL);
```

```python
%%sql
SELECT BOOSTER_VERSION, LAUNCH_SITE, LANDING_OUTCOME
FROM SPACEXTBL
WHERE LANDING_OUTCOME LIKE '%Drone Ship%'
AND YEAR(DATE) = 2015
AND LANDING_OUTCOME NOT LIKE '%Success%';
```

```python
%%sql
SELECT BOOSTER_SERIAL, SUM(PAYLOAD_MASS__KG__) AS TOTAL_MASS
FROM SPACEXTBL
WHERE BOOSTER_SERIAL IN ('B0005','B0007','B0012')
GROUP BY BOOSTER_SERIAL;
```

```python
%%sql
SELECT BOOSTER_VERSION,
AVG(CASE WHEN LANDING_OUTCOME LIKE '%Success%' THEN 1 ELSE 0 END) AS SUCCESS_RATE
FROM SPACEXTBL
GROUP BY BOOSTER_VERSION
ORDER BY SUCCESS_RATE DESC;
```
